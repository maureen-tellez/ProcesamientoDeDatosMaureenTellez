```python
import random
```

```python
## Clase "Individuo" representa a un único candidato dentro de la población

class Individuo:
    def __init__(self, longitud_cromosoma=10):
        self.cromosoma = self._generar_cromosoma(longitud_cromosoma)
        self.aptitud = 0.0  # fitness — qué tan buena es su solución

    def _generar_cromosoma(self, longitud):
        return [random.randint(0, 1) for _ in range(longitud)]

    def calcular_aptitud(self):
       
        self.aptitud = sum(self.cromosoma)
        return self.aptitud

    def mutar(self, tasa_mutacion=0.01):
    
        for i in range(len(self.cromosoma)):
            if random.random() < tasa_mutacion:
                self.cromosoma[i] = 1 - self.cromosoma[i]  # flip del bit

    def __repr__(self):
        return f"Individuo(cromosoma={self.cromosoma}, aptitud={self.aptitud})"
```

```python
## Clase "Poblacion" agrupa a todos los individuos y aplica los operadores evolutivos

class Poblacion:
    def __init__(self, tamanio=10, longitud_cromosoma=10):
        self.tamanio = tamanio
        self.generacion = 0
        self.individuos = [Individuo(longitud_cromosoma) for _ in range(tamanio)]

    def evaluar(self):
        for individuo in self.individuos:
            individuo.calcular_aptitud()

    def seleccionar_mejor(self):
        return max(self.individuos, key=lambda ind: ind.aptitud)

    def cruzar(self, padre, madre):
        punto_corte = random.randint(1, len(padre.cromosoma) - 1)
        hijo = Individuo.__new__(Individuo)
        hijo.cromosoma = padre.cromosoma[:punto_corte] + madre.cromosoma[punto_corte:]
        hijo.aptitud = 0.0
        return hijo

    def nueva_generacion(self, tasa_mutacion=0.01):
        self.evaluar()
        nueva_poblacion = []
        for _ in range(self.tamanio):
            padre = random.choice(self.individuos)
            madre = random.choice(self.individuos)
            hijo = self.cruzar(padre, madre)
            hijo.mutar(tasa_mutacion)
            nueva_poblacion.append(hijo)
        self.individuos = nueva_poblacion
        self.generacion += 1

    def __repr__(self):
        return f"Poblacion(tamanio={self.tamanio}, generacion={self.generacion})"
```

```python
## Prueba 
pop = Poblacion(tamanio=5, longitud_cromosoma=8)
pop.evaluar()
print(pop)
print("Mejor individuo:", pop.seleccionar_mejor())
pop.nueva_generacion()
print("Después de 1 generación:", pop)
```
