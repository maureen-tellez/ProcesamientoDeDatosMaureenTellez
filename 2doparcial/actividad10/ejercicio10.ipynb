```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

```python
celsius = np.array([-40,-10,0,6,15,22,38], dtype=float)
fahrenheit = np.array([-40,14,32,46,59,72,100], dtype=float)
```

```python
plt.scatter(celsius, fahrenheit)

plt.title('Correlación entre Celsius y Fahrenheit')
plt.xlablel('Celsius')
plt.show()
```

```python
X_train = celsius
y_train = fahrenheit
```

```python
print("Creando el modelo")
model = tf.keras.Sequential()
model.add(tf.keras.layers.Input(shape=(1,)))
model.add(tf.keras.layers.Dense(units=1))
```

```python
print("Complilando el modelo")
model.compile(optimizer=tf.keras.optimizers.Adam(1.0), loss='mean_squared_error')
```

```python
print("Entrenando el modelo")
epochs_hist= model.fit(X_train, y_train, epochs=100)
```

```python
# Evaluación del modelo
print("Evaluando el modelo entrenado")
print("Keys:")
print(epochs_hist.history.keys())

# Gráfico
plt.plot(epochs_hist.history['loss'])
plt.title('Progreso de Pérdida durante Entrenamiento del Modelo')
plt.xlabel('Epoch')
plt.ylabel('Training Loss')
plt.legend(['Training Loss'])
plt.show()
```

```python
# Realizar una predicción utilizando el modelo entrenado
Temp_C = 100
Temp_C_array = np.array([[Temp_C]])
Temp_F = model.predict(Temp_C_array)

print("Temperatura de Prediccion: " + str(Temp_F))
```

```python
Temp_F = 9/5 * Temp_C + 32
print("Temperatura de Ecuacion: " + str(Temp_F))
```

```python
import seaborn as sns
import pandas as pd
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

df_temperatura = pd.read_csv("celsius_a_fahrenheit.csv")
df_temperatura.head()
```

```python
print(df_temperatura.head(5))
```

```python
sns.scatterplot(x='Celsius', y='Fahrenheit', data=df_temperatura)

plt.title("Correlación entre Celsius y Fahrenheit")
plt.xlabel("Celsius")
plt.ylabel("Fahrenheit")

plt.show()
```

## Modelo Capa 1 ##

```python
#entrenando los datos 
X_train = df_temperatura['Celsius'].to_numpy(dtype=np.float32)
y_train = df_temperatura['Fahrenheit'].to_numpy(dtype=np.float32)
```

```python
#creando el modelo

model1 = tf.keras.Sequential()
model1.add(tf.keras.layers.Input(shape=(1,)))
model1.add(tf.keras.layers.Dense(units=1))
```

```python
print("Complilando el modelo")
model1.compile(optimizer=tf.keras.optimizers.Adam(1.0), loss='mean_squared_error')
```

```python
print("Entrenando el modelo")
epochs_hist1= model1.fit(X_train, y_train, epochs=100)
```

```python
# Gráfica pérdida modelo 1
plt.plot(epochs_hist1.history['loss'])
plt.title("Pérdida del Modelo (1 capa)")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
```

## Modelo Capa 2 ##

```python
oculta1 = tf.keras.layers.Dense(units=3, input_shape=[1])
oculta2 = tf.keras.layers.Dense(units=3)
salida = tf.keras.layers.Dense(units=1)
```

```python
model_2 = tf.keras.Sequential([oculta1, oculta2, salida])
```

```python
model_2.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.1),
    loss='mean_squared_error'
)
```

```python
epochs_hist_2 = model_2.fit(
    X_train,
    y_train,
    epochs=60,
    verbose=False
)
```

```python
plt.plot(epochs_hist_2.history['loss'])
plt.title("Pérdida del Modelo (capas adicionales)")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
```

```python
plt.plot(epochs_hist1.history['loss'], label="1 capa")
plt.plot(epochs_hist_2.history['loss'], label="capas extra")

plt.title("Comparación de pérdidas")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.show()
```

### Preguntas ###

* **El optimizador de algoritmo utilizado es Adam, pero existen otros como SGD y RMSProp. Describe cada uno de ellos y explica cómo pudieran cambiar los resultados de tu modelo.**
    - SGD:Actualiza los parámetros del modelo utilizando un solo ejemplo o pequeños grupos de datos en cada iteración. Es uno de los métodos más simples para entrenar redes neuronales. Sin embargo, puede ser más lento para encontrar una buena solución y suele requerir más épocas para que el modelo aprenda correctamente.
      
    - RMSProp :Ajusta automáticamente la tasa de aprendizaje de cada parámetro del modelo. Esto se hace utilizando un promedio de los gradientes anteriores, lo que ayuda a que el entrenamiento sea más estable. Gracias a esto, el modelo puede aprender de manera más rápida y con menos oscilaciones en la función de pérdida.
  
    - Adam : Utiliza promedios de los gradientes para adaptar la tasa de aprendizaje durante el entrenamiento. Por esta razón, suele ser más rápido y estable al entrenar modelos de aprendizaje automático.


* **Describe porque es importante agregar una función de pérdida como 'mean_squared_error ', investiga que otra función de pérdida se puede implementar y como alteraría los resultados.**
  
    Permite medir que tan lejos estan las predicciones del modelo respecto a los valores reales. Durante el entrenamiento, el modelo intenta minimizar esta función para mejorar su precisión.
    Otra función que se puede utilizar es mean_absolute_error (MAE), la cual calcula el promedio del valor absoluto de los errores. Si se utiliza MAE en lugar de MSE, el entrenamiento del modelo puede ser más estable, pero los resultados pueden variar ligeramente dependiendo de los datos utilizados.

**En las modificaciones realizadas,  cuantas epocas fueron 'suficientes' para que el modelo se considere entrenado para ponerlo en funcionamiento.**

  En el primer ejercicio se utilizaron 100 epochs para entrenar el modelo con una sola capa. Posteriormente, en el modelo con capas adicionales, se utilizaron 60 epochs. Al observar las gráficas de la función de pérdida, se puede notar que el modelo aprende la relación entre Celsius y Fahrenheit antes de completar todas las epochs. Esto indica que alrededor de 60 epochs fueron suficientes para que el modelo se considere entrenado y pueda realizar predicciones correctamente.
  
**¿Por qué son importantes las epoch?**
Representan el número de veces que el modelo pasa por todo el conjunto de datos durante el entrenamiento. Son importantes porque permiten que el modelo ajuste gradualmente sus parámetros para mejorar sus predicciones.




- **Que función de activación utiliza el modelo? ¿Es posible cambiarla? Si es así, cambia a una función RELU y muestra el resultado.**

Utiliza una función de activación lineal, ya que no se especifica ninguna función de activación en la capa Dense. Esto es adecuado para este problema porque la relación entre Celsius y Fahrenheit es lineal.

Se puede utilizar la función ReLU ,que es muy comun en redes neuronales porque ayuda a que el modelo aprenda relaciones no lineales.

```python
model_relu = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(3, activation='relu'),
    tf.keras.layers.Dense(1)
])

model_relu.compile(optimizer=tf.keras.optimizers.Adam(0.1),
                   loss='mean_squared_error')

history_relu = model_relu.fit(X_train, y_train, epochs=100, verbose=False)

plt.plot(history_relu.history['loss'])
plt.title("Pérdida del modelo con ReLU")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
```
