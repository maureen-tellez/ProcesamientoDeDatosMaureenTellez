```python
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
comisiones = [22,13,33,38,23,47,29,38,41,32,20,13,47,
              38,44,29,38,37,14,34,25,27,25,43,34]

llamadas = [139,132,144,142,142,142,138,139,144,134,135,137,146,
            146,144,147,144,149,131,144,132,132,127,154,147]

millas = [2371,2226,2731,3351,2289,3449,3114,3342,2842,2625,2121,2219,3463,
          3290,3103,2122,2791,3209,2287,2848,2690,2933,2671,2988,2829]

df = pd.DataFrame({
    "Comisiones": comisiones,
    "Llamadas": llamadas,
    "Millas": millas
})
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.scatter(df["Llamadas"], df["Millas"], df["Comisiones"])

ax.set_xlabel("Llamadas")
ax.set_ylabel("Millas recorridas")
ax.set_zlabel("Comisiones (miles USD)")
ax.set_title("Correlación múltiple: Comisiones vs Llamadas y Millas")

plt.show()

X = df[["Llamadas", "Millas"]]
y = df["Comisiones"]

modelo = LinearRegression()
modelo.fit(X, y)

R_cuadrada = modelo.score(X, y)
R_multiple = np.sqrt(R_cuadrada)

R_multiple, R_cuadrada
```

```python
coef_table = pd.DataFrame({
    "Variable": ["Intercepto", "Llamadas", "Millas"],
    "Coeficiente": [modelo.intercept_, modelo.coef_[0], modelo.coef_[1]]
})

coef_table
```

**Conslusión**

El modelo indica que tanto las llamadas como las millas recorridas influyen positivamente en las comisiones. 
Cada llamada adicional incrementa las comisiones en aproximadamente 0.63 dólares, mientras que cada milla recorrida incrementa las comisiones en aproximadamente 0.0157  dólares.
Esto confirma que ambas variables contribuyen al desempeño en ventas.
