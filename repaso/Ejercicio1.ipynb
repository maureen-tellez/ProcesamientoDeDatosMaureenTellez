```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Ingreso": [1452,1361,1426,1470,1456,1430,1354,1442,1394,1459,1399,1458,1537,
                1425,1445,1439,1348,1450,1431,1446,1485,1405,1461,1490,1426],
    "Habitaciones": [23,47,21,39,37,29,23,44,45,16,30,42,54,
                     27,34,15,19,38,44,47,43,38,51,61,39]
}

df = pd.DataFrame(data)

# Correlación de Pearson
corr = df.corr().iloc[0,1]

# Gráfica de dispersión
plt.figure()
plt.scatter(df["Habitaciones"], df["Ingreso"])
plt.xlabel("Habitaciones ocupadas")
plt.ylabel("Ingreso restaurante")
plt.title("Gráfica de dispersión: Habitaciones vs Ingreso")
plt.show()
```

```python
corr
```

En la grafica es de tendencia ascendente, lo que indica que conforme aumenta el número de llamadas realizadas por los agentes, las comisiones tienden a incrementarse. 
Los puntos no están completamente alineados, sino que presentan cierta dispersión. Esto significa que, aunque existe una relación positiva entre ambas variables, no es perfecta.
